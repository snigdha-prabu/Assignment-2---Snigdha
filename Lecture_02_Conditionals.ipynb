## Why Conditionals?
Programs often need to make decisions. Conditionals let the program take different paths depending on data or user input.

### Basic `if` Statement

```python
x = 5
if x > 0:
    print("x is positive")
else:
    print("x is not positive")
```

### Truthiness in Python
Values that evaluate to `False`: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `None`.
Everything else is `True`.

```python
values = [0, 0.0, ""," ", [], {}, (), None, 42, "hello"]
for v in values:
    if v:
        print(f"{v!r} is True")
    else:
        print(f"{v!r} is False")
```

### `if-else` Structure

```python
name = input("What is your name?")
if name == "Greg":
    print("Welcome, esteemed Dr. Cousins!")
else:
    print("Hello, regular person", name)
```

### `if-elif-else` Chain

```python
#This is the score
score =79

if score >= 90: #first condition
    print("A")
elif score >= 80: #second condition
    print("B")
elif score >= 70: #third condition
    print("C")
elif score >= 60:
    print("D")
elif score >=50:
    print("Below D")    
else: 
    print("YOU HAVE FAILED")#what is done if no conditions are met/True.
```

### Chained Comparisons

```python
x = 1.5
if 0 <= x <= 10:
    print("x is between 1 and 10")
else: 
    print("x is NOT between 1 and 10")
```

### Example: Leap Year Test

```python
year = 2024
if year % 4 != 0: # != means "not equal to"
    print("Not a leap year")
elif year % 100 != 0:
    print("Leap year")
elif year % 400 != 0:
    print("Not a leap year")
else:
    print("Leap year")
```
